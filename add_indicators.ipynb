```python
import pandas as pd
import pandas_ta as ta

data_dir = "../data/stocks"
stock_names = ["A","AA","AACG","AAL","AAMC"]
new_data_dir = "./"

for stock_name in stock_names:
    stock_data = pd.read_csv(f"{data_dir}/{stock_name}.csv")                  
    
    stock_data.ta.rsi(close='Close', length=14, append=True)
    stock_data['RSI_signal'] = 0
    stock_data.loc[stock_data['RSI_14'] < 30,"RSI_signal"] = 1
    stock_data.loc[stock_data['RSI_14'] > 70,"RSI_signal"] = -1
    
    
    stock_data['MA_short'] = stock_data['Close'].rolling(window=5).mean()
    stock_data['MA_long'] = stock_data['Close'].rolling(window=20).mean()
    stock_data['MA_signal'] = 0
    stock_data.loc[stock_data['MA_short'] > stock_data['MA_long'],"MA_signal"] = 1
    stock_data.loc[stock_data['MA_short'] > stock_data['MA_long'],"MA_signal"] = -1
    
    stock_data.to_csv(f"./stocks/{stock_name}.csv", index=False)
    
```
